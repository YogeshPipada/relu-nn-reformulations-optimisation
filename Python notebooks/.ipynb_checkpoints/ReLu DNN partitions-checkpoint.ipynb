```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed = 12
layer_sizes = np.array([2, 4, 4, 4])
layer_weights = {}
for idx, layer_size in list(enumerate(layer_sizes))[:-1]:
    input_size = layer_size
    output_size = layer_sizes[idx+1]
    layer_weights[idx] = {}
    layer_weights[idx]["weights"] = np.random.uniform(-1, 1, size=(input_size, output_size))
    layer_weights[idx]["bias"] = np.random.uniform(-0.3, 0.3, size=(output_size))
```

```python
x_1 = np.round(np.linspace(-1, 1, 20), 2)
x_2 = np.round(np.linspace(-1, 1, 20), 2)
X_1, X_2 = np.meshgrid(x_1, x_2) 
layer_output = {}
```

```python
# layer 1 output 
weights = layer_weights[0]["weights"]
bias = layer_weights[0]["bias"]
Z_list = []
activations_list = []
for idx, row in enumerate(weights.transpose()):
    Z = row[0] * X_1 + row[1] * X_2 + bias[idx]
    Z_list.append(Z)
    activations = (Z > 0).astype(int)
    activations_list.append(activations)
    plt.contour(X_1, X_2, Z, levels=[0])

layer_output[1] = {"outputs": Z_list, "activations": activations_list}
```

```python
# Layer 1 output
weights = layer_weights[1]["weights"]
bias = layer_weights[1]["bias"]
outputs = layer_output[1]["outputs"]
activations = layer_output[1]["activations"]
for idx, row in enumerate(weights.transpose()):
    Z = row[0] * activations[0] * outputs[0] + bias[idx]
    activations = (Z > 0).astype(int)
    layer_output[idx] = {"outputs": Z, "activations": activations}
    plt.contour(X_1, X_2, Z, levels=[0])
```
