```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed = 17
layer_sizes = np.array([2, 4, 4, 4])
layer_weights = {}
for idx, layer_size in list(enumerate(layer_sizes))[:-1]:
    input_size = layer_size
    output_size = layer_sizes[idx+1]
    layer_weights[idx] = {}
    layer_weights[idx]["weights"] = np.random.uniform(-3,  3, size=(input_size, output_size))
    layer_weights[idx]["bias"] = np.random.uniform(-1, 1, size=(output_size))
```

```python
x_1 = np.round(np.linspace(-10, 10, 1000), 2)
x_2 = np.round(np.linspace(-10, 10, 1000), 2)
X_1, X_2 = np.meshgrid(x_1, x_2) 
```

```python
# Input layer output
weights = layer_weights[0]["weights"]
bias = layer_weights[0]["bias"]
Z_list = []
activation_list = []
for idx, row in enumerate(weights.transpose()):
    Z = row[0] * X_1 + row[1] * X_2 + bias[idx]
    # Z = (Z - np.min(Z)) / (np.max(Z) - np.min(Z))
    activations = (Z > 0).astype(int)
    Z_list.append(Z)
    activation_list.append(activations)
    plt.contour(X_1, X_2, Z, levels=[0], cmap="viridis")

weights = layer_weights[1]["weights"]
bias = layer_weights[1]["bias"]
output = Z_list
u = activation_list
Z_list2 = []
activation_list2 = []
for idx, row in enumerate(weights.transpose()):
    Z = (row[0] * output[0] * u[0] + row[1] * output[1] * u[1] + row[2] * output[2] * u[2] + row[3] * output[3] * u[3]+ bias[idx])
    # Z = (Z - np.min(Z)) / (np.max(Z) - np.min(Z))
    Z_list2.append(Z)
    activation_list2.append(activations)
    plt.contour(X_1, X_2, Z, levels=[0], cmap="Wistia_r")

weights = layer_weights[2]["weights"]
bias = layer_weights[2]["bias"]
output = Z_list2
u = activation_list2
for idx, row in enumerate(weights.transpose()):
    Z = (row[0] * output[0] * u[0] + row[1] * output[1] * u[1] + row[2] * output[2] * u[2] + row[3] * output[3] * u[3]+ bias[idx])
    # Z = (Z - np.min(Z)) / (np.max(Z) - np.min(Z))
    plt.contour(X_1, X_2, Z, levels=[0], cmap="Greens_r")
```
