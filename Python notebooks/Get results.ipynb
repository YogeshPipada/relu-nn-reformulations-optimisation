```python
import pandas as pd
from matplotlib import pyplot as plt
import os
import sys
import numpy as np
```

```python
current_path = os.getcwd()
cwd = os.path.dirname(current_path)
sys.path.append(os.path.join(cwd, "utilities"))

# Import pickle operations

from pickle_file_operations import read_pickle, write_pickle
from optimization_dataclasses import Optimization_Results
```

```python
def my_func(x, y, a, b):
    x_new = max(0.098, x)
    y_new = max(0.10, y)
    if y_new / x_new > 1:
        z = (-1 * b) ** -1 * x_new * ((np.log((y_new / x_new) - 1)) - a)
    else:
        x_new = 0.99 * y_new
        z = (-1 * b) ** -1 * x_new * ((np.log((y_new / x_new) - 1)) - a)
        
    z_positive = max(z, 0)
    return round(z_positive, 3)
    
def calculate_flexibility_cost(x, y, shaping_parameters):
    a, b = shaping_parameters
    return my_func(x, y, a, b)
```

```python
results_path = os.path.join(cwd, "results")
input_path = os.path.join(cwd, "optimization input")
pwl_input = read_pickle(os.path.join(input_path, "pwl_data_input.pickle"))
folders = os.listdir(results_path)
```

```python

dict_data = {}
for folder in folders:
    print(folder + "\n")
    dict_data[folder] = {}
    filepath = os.path.join(results_path, folder)
    for filename in os.listdir(filepath):
        print(filename + "\n")
        data = read_pickle(os.path.join(filepath, filename))
        obj_val = 0
        for time in range(24):
            x = data["flexibility_MWh"][time]
            y = data["updated_max_flexibility_MWh"][time]
            flexibility_cost = calculate_flexibility_cost(x, y, pwl_input[time].shaping_parameters)
            revenue_generated = x * data["Market_price"][time]
            obj_val = obj_val + revenue_generated - flexibility_cost
        print(obj_val)
    
```
